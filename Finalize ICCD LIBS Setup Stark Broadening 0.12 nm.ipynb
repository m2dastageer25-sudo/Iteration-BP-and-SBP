```python
# ============================================================
# SAHA-BOLTZMANN ITERATION
# RELATIVE-ERROR UNCERTAINTY METHOD — LINEAR / WORST-CASE
#
# 10 Ca I lines + 6 Ca II lines = 16 lines
#
# Initial temperature:
#     Obtained automatically from the 10 Ca I Boltzmann plot
#
# Saha correction:
#
# Y = ln(I lambda / h c A_k g_k)
#     - z ln[ 2(2*pi*me*kB*T)^(3/2) / (h^3 Ne) ]
#
# Electron density:
#
# Ne = Delta_lambda / (2 omega) * N_reference
#
# Stark width:
#
# Delta_lambda = 0.170 +/- 0.016 nm
#
# UNCERTAINTY METHOD:
#
# Slope:
#     sigma_m from linear regression
#
# Temperature:
#     sigma_T / T = sigma_m / |m|
#
# Electron density (WORST-CASE / MAXIMUM RELATIVE-ERROR METHOD):
#     Delta_lambda and omega are independent quantities, so their
#     relative uncertainties are ADDED (not combined in quadrature).
#
#     sigma_omega (ROUTE A - worst case, no cancellation):
#     omega(T) = a + bT - cT^2 is a SUM of terms with different
#     powers of T. Each T-dependent term's own uncertainty
#     contribution is taken separately (as if independent) and
#     the magnitudes are ADDED - the two terms are NOT allowed
#     to partially cancel, even though they share the same T:
#
#     sigma_omega = ( |b| + |2cT| ) * sigma_T
#
#     (Route B, which combines b and 2cT algebraically BEFORE
#     taking the absolute value - i.e. sigma_omega = |b-2cT|*sigma_T -
#     is the more rigorous/correlated treatment, but Route A is
#     used here for consistency with the worst-case convention
#     applied everywhere else in this notebook.)
#
#     sigma_Ne/Ne = sigma_lambda/lambda + sigma_omega/omega
#
# No previous files are used.
# ============================================================


import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. Ca I DATA
# ============================================================


# x-axis Ek (cm-1) upper energy level
E_CaI = np.array([
    42170,
    42171,
    39335,
    38259,
    38192,
    31539,
    31539,
    31539,
    36574,
    37751
], dtype=float)

# y-axis using  "ln(λI/Akgkhc)"
Y_CaI = np.array([
   36.71524242374016,
   36.45823589572427,
   38.6988268462164,
   35.39728348286384,
   38.66038204239678,
   38.164880732560285,
   37.49093675542158,
   37.32986956924025,
   37.28391530019969,
   37.92086665815769
], dtype=float)


# ============================================================
# 2. Ca II DATA
# ============================================================
# x-axis Ek (cm-1) upper energy level
E_CaII = np.array([
    106139,
    106158,
    101466,
    101466,
    74714,
    74491
], dtype=float)

# y-axis Raw mean without saha correction using "ln(λI/Akgkhc)"
Y_CaII_raw = np.array([
    34.997270710599686,
    34.471736599356,
    36.90682177959408,
    36.271286143452826,
    34.72136086676374,
    35.361575369785704
], dtype=float)


# ============================================================
# 2b. LINE-BY-LINE UNCERTAINTY INPUTS
# ============================================================
# Instrumental wavelength uncertainty: 0.05 nm for every line.
# Relative wavelength uncertainty is therefore 0.05/lambda.
# Transition-probability uncertainty follows the user's grade
# convention: C = 25%, D = 50%.
# Intensity uncertainty values are the previously established
# relative uncertainties for each line.

# Ca I: 10 lines
wavelength_CaI_nm = np.array([
    458.14, 458.58, 526.17, 558.19, 560.28,
    610.27, 612.22, 616.21, 616.95, 445.58
], dtype=float)

deltaI_over_I_CaI = np.array([
    0.000140422,
    9.11015E-05,
    5.18999E-06,
    0.000560656,
    3.99962E-05,
    1.99149E-05,
    1.05741E-05,
    5.76031E-05,
    5.33925E-05,
    9.6878E-06
], dtype=float)

# Ca I transition grades: C, C, D, D, D, C, C, C, D, C
grade_CaI = np.array(['C','C','D','D','D','C','C','C','D','C'])
deltaA_over_A_CaI = np.where(grade_CaI == 'D', 0.50, 0.25)

# Ca II: 6 lines -- all are C grade in the revised table.
wavelength_CaII_nm = np.array([
    315.89, 317.93, 370.60, 373.69, 393.36, 396.85
], dtype=float)

deltaI_over_I_CaII = np.array([
    0.000110052,
    6.59258E-06,
    1.23403E-05,
    6.08059E-05,
    2.37766E-05,
    0.000106224
], dtype=float)

grade_CaII = np.array(['C','C','C','C','C','C'])
deltaA_over_A_CaII = np.where(grade_CaII == 'D', 0.50, 0.25)

# Fixed instrumental wavelength uncertainty
delta_lambda_instrument = 0.50  # nm

# Relative wavelength uncertainties, line by line
deltaLambda_over_lambda_CaI = (
    delta_lambda_instrument / wavelength_CaI_nm
)
deltaLambda_over_lambda_CaII = (
    delta_lambda_instrument / wavelength_CaII_nm
)


# ============================================================
# 2c. LINEAR RELATIVE-UNCERTAINTY PROPAGATION
# ============================================================
# Boltzmann Ca I y-error:
#   Delta y = Delta I/I + Delta lambda/lambda + Delta A/A
#
# Saha-Boltzmann Ca II y-error:
#   Delta y = Delta I/I + Delta lambda/lambda + Delta A/A
#              + Delta S/S
#
# where, for the Saha factor S,
#   Delta S/S = (3/2)(Delta T/T) + Delta Ne/Ne
#
# All independent relative contributions are ADDED linearly,
# NOT combined in quadrature and NOT estimated by half-difference.
# ============================================================

def line_y_uncertainty(deltaI_over_I, deltaLambda_over_lambda, deltaA_over_A):
    return (
        deltaI_over_I
        + deltaLambda_over_lambda
        + deltaA_over_A
    )


sigma_Y_CaI_propagated = line_y_uncertainty(
    deltaI_over_I_CaI,
    deltaLambda_over_lambda_CaI,
    deltaA_over_A_CaI
)




# ============================================================
# 3. PHYSICAL CONSTANTS
# ============================================================

h = 6.62607015e-34       # J s
kB = 1.380649e-23        # J/K
me = 9.1093837e-31       # kg

# hc/kB when E is in cm^-1
hc_over_k = 1.4387769    # cm K


# ============================================================
# 4. STARK WIDTH
# ============================================================

delta_lambda = 0.12
sigma_delta_lambda = 0.05

# nm

# Reference electron density
N_reference = 1.0e16     # cm^-3


# ============================================================
# 5. OMEGA(T)
# ============================================================

def omega_function(T):

    omega = (
        4.8767e-4
        + 1.6385e-8 * T
        - 1.8473e-13 * T**2
    )

    return omega


# ============================================================
# 6. ELECTRON DENSITY
# ============================================================

def calculate_Ne(T):

    omega = omega_function(T)

    Ne = (
        delta_lambda
        /
        (2.0 * omega)
        *
        N_reference
    )

    return omega, Ne


# ============================================================
# 7. SAHA CORRECTION
#
# Your equation:
#
# Y_coordinate =
#
# ln(I lambda / h c A_k g_k)
#
# -
#
# z ln[
#     2(2*pi*me*kB*T)^(3/2)
#     -----------------------
#             h^3 Ne
# ]
#
# For Ca II:
#
# z = 1
# ============================================================

def calculate_saha_factor(T, Ne):

    # Ne is supplied in cm^-3; convert to m^-3 for SI constants.
    Ne_m3 = Ne * 1.0e6

    S = (
        2.0
        * (2.0 * np.pi * me * kB * T)**1.5
        / (h**3 * Ne_m3)
    )

    return S


def calculate_saha_correction(T, Ne, z=1):
    # The Saha-Boltzmann coordinate contains -z*ln(S).
    S = calculate_saha_factor(T, Ne)
    return z * np.log(S)


# ============================================================
# 7b. SAHA FACTOR UNCERTAINTY
# ============================================================
# S = 2(2*pi*m_e*k*T)^(3/2)/(Ne*h^3)
# Therefore, using the requested LINEAR relative-error method:
#
# Delta S/S = (3/2)(Delta T/T) + Delta Ne/Ne
#
# Because ln(S) appears in the y-coordinate,
# Delta[ln(S)] = Delta S/S.
# For Ca II, z = 1, so the Saha contribution to Delta y is
# numerically equal to Delta S/S.
# ============================================================

def calculate_saha_uncertainty(T, sigma_T, Ne, sigma_Ne, z=1):

    relative_S_error = (
        1.5 * sigma_T / T
        + sigma_Ne / Ne
    )

    return z * relative_S_error


# ============================================================
# 8. LINEAR FIT
#
# Y = mE + b
#
# Also calculate slope uncertainty.
# ============================================================

def linear_fit(E, Y):

    N = len(E)

    # Linear regression
    m, b = np.polyfit(E, Y, 1)

    # Fitted Y values
    Y_fit = m * E + b

    # Residuals
    residuals = Y - Y_fit

    # Residual sum of squares
    SS_res = np.sum(residuals**2)

    # Degrees of freedom
    dof = N - 2

    # Residual variance
    s2 = SS_res / dof

    # Spread in E
    Sxx = np.sum((E - np.mean(E))**2)

    # Standard uncertainty of slope
    sigma_m = np.sqrt(s2 / Sxx)

    # Intercept uncertainty
    sigma_b = np.sqrt(
        s2 *
        (
            1.0 / N
            +
            np.mean(E)**2 / Sxx
        )
    )

    # R squared
    SS_tot = np.sum(
        (Y - np.mean(Y))**2
    )

    R2 = 1.0 - SS_res / SS_tot

    return (
        m,
        b,
        sigma_m,
        sigma_b,
        Y_fit,
        R2
    )


# ============================================================
# 8b. FIT-LINE UNCERTAINTY BAND (WORST CASE)
#
# Worst-case upper/lower bounds of the regression line,
# obtained by shifting slope and intercept by their maximum
# (+/- 1 sigma) uncertainties simultaneously in the direction
# that maximises the spread:
#
# Y_upper(E) = (m + sigma_m) E + (b + sigma_b)
# Y_lower(E) = (m - sigma_m) E + (b - sigma_b)
# ============================================================

def fit_uncertainty_band(E_range, m, sigma_m, b, sigma_b):

    Y_upper = (m + sigma_m) * E_range + (b + sigma_b)
    Y_lower = (m - sigma_m) * E_range + (b - sigma_b)

    return Y_lower, Y_upper


# ============================================================
# 9. TEMPERATURE FROM SLOPE
#
# m = -1.4387769 / T
#
# Therefore:
#
# T = -1.4387769 / m
# ============================================================

def temperature_from_slope(m):

    T = -hc_over_k / m

    return T


# ============================================================
# 10. TEMPERATURE UNCERTAINTY
#
# RELATIVE ERROR METHOD
#
# sigma_T / T = sigma_m / |m|
#
# Therefore:
#
# sigma_T = T * sigma_m / |m|
# ============================================================

def temperature_uncertainty(T, m, sigma_m):

    relative_error_T = (
        sigma_m / abs(m)
    )

    sigma_T = (
        T * relative_error_T
    )

    return sigma_T


# ============================================================
# 11. ELECTRON DENSITY UNCERTAINTY
#
# WORST-CASE (MAXIMUM) RELATIVE-ERROR METHOD
#
# Ne = Delta_lambda / (2 omega) * N_reference
#
# Delta_lambda (Stark width) and omega (Stark broadening
# parameter) are two INDEPENDENT measured/tabulated quantities.
# Because their individual errors could in principle combine in
# the worst possible (reinforcing) direction, the relative
# uncertainties are ADDED directly (not combined in quadrature):
#
# sigma_Ne/Ne = sigma_lambda/lambda + sigma_omega/omega
#
# Note: omega(T) = a + bT - cT^2 is a SUM of terms with
# different powers of the SAME variable, T. Per the worst-case
# convention used throughout this notebook, the linear (bT) and
# quadratic (-cT^2) terms' contributions to sigma_omega are
# taken separately and ADDED (magnitudes, no cancellation):
#
# sigma_omega = ( |b| + |2cT| ) * sigma_T   (see Section 6 below)
# ============================================================

def electron_density_uncertainty(
    T,
    sigma_T,
    omega,
    Ne
):

    # --------------------------------------------------------
    # Relative uncertainty from Stark width
    # --------------------------------------------------------

    relative_lambda_error = (
        sigma_delta_lambda
        /
        delta_lambda
    )


    # --------------------------------------------------------
    # Uncertainty in omega -- ROUTE A (worst case, no cancellation)
    #
    # omega(T) = a + bT - cT^2
    #
    # The linear term (bT) and the quadratic term (-cT^2) are
    # each treated as separate, independent contributors to the
    # uncertainty in omega. Their magnitudes are ADDED directly
    # (worst case) -- they are NOT combined algebraically first
    # (that would be Route B, which allows partial cancellation
    # since both terms share the same underlying T).
    #
    # sigma_omega = ( |b| + |2cT| ) * sigma_T
    # --------------------------------------------------------

    b_coeff = 1.6385e-8
    c_coeff = 1.8473e-13

    sigma_omega = (
        abs(b_coeff)
        +
        abs(2.0 * c_coeff * T)
    ) * sigma_T


    # --------------------------------------------------------
    # Relative omega uncertainty
    # --------------------------------------------------------

    relative_omega_error = (
        sigma_omega
        /
        abs(omega)
    )


    # --------------------------------------------------------
    # Total relative uncertainty in Ne
    #
    # WORST-CASE (MAXIMUM) RELATIVE-ERROR METHOD:
    # relative errors of independent quantities (lambda, omega)
    # are ADDED, not combined in quadrature, since we assume the
    # worst-case scenario in which both errors reinforce each
    # other (no cancellation assumed).
    # --------------------------------------------------------

    relative_Ne_error = (
        relative_lambda_error
        +
        relative_omega_error
    )


    # --------------------------------------------------------
    # Absolute uncertainty in Ne
    # --------------------------------------------------------

    sigma_Ne = (
        Ne
        *
        relative_Ne_error
    )


    return (
        sigma_Ne,
        sigma_omega,
        relative_lambda_error,
        relative_omega_error,
        relative_Ne_error
    )


# ============================================================
# 12. INITIAL Ca I BOLTZMANN PLOT
# ============================================================

(
    m_initial,
    b_initial,
    sigma_m_initial,
    sigma_b_initial,
    Yfit_initial,
    R2_initial
) = linear_fit(
    E_CaI,
    Y_CaI
)


# Initial temperature
T_initial = temperature_from_slope(
    m_initial
)


# Initial temperature uncertainty
sigma_T_initial = temperature_uncertainty(
    T_initial,
    m_initial,
    sigma_m_initial
)

# ------------------------------------------------------------
# Propagated Ca I y-error bars
#
# Delta y_CaI = Delta I/I + Delta lambda/lambda + Delta A/A
# using the requested linear relative-error method.
# ------------------------------------------------------------
sigma_Y_CaI = sigma_Y_CaI_propagated.copy()


# ============================================================
# 13. PRINT INITIAL RESULT
# ============================================================

print()
print("=" * 90)
print("INITIAL Ca I BOLTZMANN PLOT")
print("=" * 90)

print(
    f"Slope = {m_initial:.9e} +/- "
    f"{sigma_m_initial:.9e} cm"
)

print(
    f"Relative slope uncertainty = "
    f"{sigma_m_initial/abs(m_initial)*100:.3f} %"
)

print(
    f"Temperature = {T_initial:.3f} +/- "
    f"{sigma_T_initial:.3f} K"
)

print(
    f"Relative temperature uncertainty = "
    f"{sigma_T_initial/T_initial*100:.3f} %"
)

print(
    f"R^2 = {R2_initial:.6f}"
)

print("=" * 90)


# ============================================================
# 14. PLOT INITIAL Ca I
# ============================================================

plt.figure(figsize=(9, 6))

# Ca I points with empirical residual-scatter y-error bars.
# These bars represent the 1-sigma residual scatter of the
# initial 10-line regression; they are not individual intensity
# measurement uncertainties.
plt.errorbar(
    E_CaI,
    Y_CaI,
    yerr=sigma_Y_CaI,
    fmt="o",
    markersize=7,
    capsize=4,
    elinewidth=1.2,
    ecolor="black",
    label=r"Ca I lines ($\pm1\sigma_Y$, residual scatter)",
    zorder=3
)

E_line = np.linspace(
    E_CaI.min() - 500,
    E_CaI.max() + 500,
    500
)

Y_line = (
    m_initial * E_line
    +
    b_initial
)

# Worst-case fit-line uncertainty band
Y_line_lower, Y_line_upper = fit_uncertainty_band(
    E_line,
    m_initial,
    sigma_m_initial,
    b_initial,
    sigma_b_initial
)

plt.fill_between(
    E_line,
    Y_line_lower,
    Y_line_upper,
    alpha=0.2,
    label="Fit uncertainty (worst case)",
    zorder=1
)

plt.plot(
    E_line,
    Y_line,
    linewidth=2,
    label="Linear fit",
    zorder=2
)

plt.xlabel(
    r"Upper-level energy $E_k$ (cm$^{-1}$)",
    fontsize=12
)

plt.ylabel(
    "Y-coordinate",
    fontsize=12
)

plt.title(
    "Initial Boltzmann Plot — 10 Ca I Lines",
    fontsize=14
)

text_initial = (
    f"T = {T_initial:.2f} ± "
    f"{sigma_T_initial:.2f} K\n"
    f"Slope = {m_initial:.3e} ± "
    f"{sigma_m_initial:.2e} cm\n"
    f"R² = {R2_initial:.5f}"
)

plt.text(
    0.97,
    0.05,
    text_initial,
    transform=plt.gca().transAxes,
    ha="right",
    va="bottom",
    bbox=dict(
        facecolor="white",
        edgecolor="black",
        alpha=0.9
    )
)

plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 15. START ITERATION
#
# IMPORTANT:
# Initial temperature comes from Ca I plot.
# We do NOT use 6851 K manually.
# ============================================================

T = T_initial
sigma_T = sigma_T_initial


# Maximum number of iterations
maximum_iterations = 11

# Convergence criterion
tolerance = 0.01


# Store results
results = []


# ============================================================
# 16. SAHA-BOLTZMANN ITERATIONS
# ============================================================

for iteration in range(maximum_iterations):


    # --------------------------------------------------------
    # A. Calculate omega and electron density
    # --------------------------------------------------------

    omega, Ne = calculate_Ne(T)


    # --------------------------------------------------------
    # B. Calculate electron-density uncertainty
    # --------------------------------------------------------

    (
        sigma_Ne,
        sigma_omega,
        relative_lambda_error,
        relative_omega_error,
        relative_Ne_error
    ) = electron_density_uncertainty(
        T,
        sigma_T,
        omega,
        Ne
    )


    # --------------------------------------------------------
    # C. Calculate Saha correction
    # --------------------------------------------------------

    saha = calculate_saha_correction(
        T,
        Ne,
        z=1
    )


    # --------------------------------------------------------
    # C2. Saha-factor relative uncertainty
    # --------------------------------------------------------

    relative_S_error = calculate_saha_uncertainty(
        T,
        sigma_T,
        Ne,
        sigma_Ne,
        z=1
    )

    sigma_saha = relative_S_error  # absolute uncertainty in ln(S)

    # --------------------------------------------------------
    # D. Correct Ca II Y values and calculate line-by-line
    # y-error bars.
    #
    # Delta y_CaII = Delta I/I + Delta lambda/lambda + Delta A/A
    #                + Delta S/S
    # --------------------------------------------------------

    Y_CaII_corrected = (
        Y_CaII_raw
        -
        saha
    )

    sigma_Y_CaII_line = line_y_uncertainty(
        deltaI_over_I_CaII,
        deltaLambda_over_lambda_CaII,
        deltaA_over_A_CaII
    )

    sigma_Y_CaII_corrected = (
        sigma_Y_CaII_line
        + relative_S_error
    )


    # --------------------------------------------------------
    # E. Combine all 16 lines
    # --------------------------------------------------------

    E_all = np.concatenate([
        E_CaI,
        E_CaII
    ])

    Y_all = np.concatenate([
        Y_CaI,
        Y_CaII_corrected
    ])


    # --------------------------------------------------------
    # F. Fit all 16 points
    # --------------------------------------------------------

    (
        m,
        b,
        sigma_m,
        sigma_b,
        Y_fit,
        R2
    ) = linear_fit(
        E_all,
        Y_all
    )


    # --------------------------------------------------------
    # G. Calculate new temperature
    # --------------------------------------------------------

    T_new = temperature_from_slope(
        m
    )


    # --------------------------------------------------------
    # H. Temperature uncertainty
    #
    # Relative error:
    #
    # sigma_T/T = sigma_m/|m|
    # --------------------------------------------------------

    sigma_T_new = temperature_uncertainty(
        T_new,
        m,
        sigma_m
    )


    # --------------------------------------------------------
    # I. Save results
    # --------------------------------------------------------

    results.append([
        iteration,
        T,
        sigma_T,
        omega,
        sigma_omega,
        Ne,
        sigma_Ne,
        saha,
        m,
        sigma_m,
        T_new,
        sigma_T_new,
        R2,
        sigma_saha
    ])


    # ========================================================
    # J. PRINT CURRENT ITERATION
    # ========================================================

    print()
    print("=" * 100)
    print(f"ITERATION {iteration}")
    print("=" * 100)

    print(
        f"T input = "
        f"{T:.3f} +/- {sigma_T:.3f} K"
    )

    print(
        f"omega = "
        f"{omega:.9e} +/- "
        f"{sigma_omega:.9e} nm"
    )

    print(
        f"Ne = "
        f"{Ne:.6e} +/- "
        f"{sigma_Ne:.6e} cm^-3"
    )

    print(
        f"Relative Ne uncertainty = "
        f"{relative_Ne_error*100:.3f} %"
    )

    print(
        f"Saha correction = "
        f"{saha:.6f} +/- {sigma_saha:.6f}"
    )

    print(
        f"Slope = "
        f"{m:.9e} +/- "
        f"{sigma_m:.9e} cm"
    )

    print(
        f"Relative slope uncertainty = "
        f"{sigma_m/abs(m)*100:.3f} %"
    )

    print(
        f"T output = "
        f"{T_new:.3f} +/- "
        f"{sigma_T_new:.3f} K"
    )

    print(
        f"Relative temperature uncertainty = "
        f"{sigma_T_new/T_new*100:.3f} %"
    )

    print(
        f"R^2 = {R2:.6f}"
    )

    print()
    print("UNCERTAINTY PROPAGATION FOR THIS ITERATION")
    print("Ca I:  Delta y = DeltaI/I + Deltalambda/lambda + DeltaA/A")
    print("Ca II: Delta y = DeltaI/I + Deltalambda/lambda + DeltaA/A + DeltaS/S")
    print(f"DeltaS/S = {relative_S_error:.6f} ({relative_S_error*100:.3f} %)" )

    print()
    print("Ca II line-by-line y-error bars")
    print("-" * 110)
    print(f"{'lambda(nm)':>11} | {'dI/I':>10} | {'dl/l':>10} | {'dA/A':>10} | {'dS/S':>10} | {'dy':>10} | {'Y':>10} | {'Y +/- dy':>18}")
    print("-" * 110)
    for i in range(6):
        print(
            f"{wavelength_CaII_nm[i]:11.2f} | "
            f"{deltaI_over_I_CaII[i]:10.6f} | "
            f"{deltaLambda_over_lambda_CaII[i]:10.7f} | "
            f"{deltaA_over_A_CaII[i]:10.2f} | "
            f"{relative_S_error:10.6f} | "
            f"{sigma_Y_CaII_corrected[i]:10.6f} | "
            f"{Y_CaII_corrected[i]:10.5f} | "
            f"{Y_CaII_corrected[i]:10.5f} +/- {sigma_Y_CaII_corrected[i]:.5f}"
        )

    print()
    print("Ca I line-by-line y-error bars")
    print("-" * 95)
    print(f"{'lambda(nm)':>11} | {'grade':>5} | {'dI/I':>10} | {'dl/l':>10} | {'dA/A':>10} | {'dy':>10} | {'Y':>10}")
    print("-" * 95)
    for i in range(10):
        print(
            f"{wavelength_CaI_nm[i]:11.2f} | "
            f"{grade_CaI[i]:>5} | "
            f"{deltaI_over_I_CaI[i]:10.6f} | "
            f"{deltaLambda_over_lambda_CaI[i]:10.7f} | "
            f"{deltaA_over_A_CaI[i]:10.2f} | "
            f"{sigma_Y_CaI[i]:10.6f} | "
            f"{Y_CaI[i]:10.5f}"
        )


    # ========================================================
    # K. PRINT CORRECTED Ca II VALUES
    # ========================================================

    print()
    print("Ca II Saha-corrected Y values")
    print("-" * 70)

    for i in range(6):

        print(
            f"{i+1:2d}. "
            f"E = {E_CaII[i]:.0f} cm^-1   "
            f"Raw Y = {Y_CaII_raw[i]:.5f}   "
            f"Corrected Y = "
            f"{Y_CaII_corrected[i]:.5f}"
        )


    # ========================================================
    # L. DRAW BOLTZMANN PLOT FOR THIS ITERATION
    #
    # The Ca I dataset is unchanged during Saha iterations, so
    # this plot is mathematically identical to the initial Ca I
    # Boltzmann plot; it is redrawn here deliberately so every
    # iteration has a complete, self-contained plot and summary.
    # ========================================================

    plt.figure(figsize=(9, 6))
    plt.errorbar(
        E_CaI,
        Y_CaI,
        yerr=sigma_Y_CaI,
        fmt="o",
        markersize=7,
        capsize=4,
        elinewidth=1.2,
        ecolor="black",
        label=r"Ca I ($\pm\Delta y$)",
        zorder=3
    )

    E_line_CaI = np.linspace(E_CaI.min() - 500, E_CaI.max() + 500, 500)
    Y_line_CaI = m_initial * E_line_CaI + b_initial
    Y_low_CaI, Y_high_CaI = fit_uncertainty_band(
        E_line_CaI, m_initial, sigma_m_initial, b_initial, sigma_b_initial
    )
    plt.fill_between(E_line_CaI, Y_low_CaI, Y_high_CaI, alpha=0.15, label="Fit uncertainty")
    plt.plot(E_line_CaI, Y_line_CaI, linewidth=2, label="Ca I linear fit")
    plt.xlabel(r"Upper-level energy $E_k$ (cm$^{-1}$)")
    plt.ylabel("Boltzmann Y-coordinate")
    plt.title(f"Boltzmann Plot — Ca I — Iteration {iteration}")
    plt.text(
        0.02, 0.97,
        f"Iteration = {iteration}\nT(Ca I) = {T_initial:.2f} ± {sigma_T_initial:.2f} K\nR² = {R2_initial:.5f}",
        transform=plt.gca().transAxes, ha="left", va="top",
        bbox=dict(facecolor="white", edgecolor="black", alpha=0.9)
    )
    plt.grid(alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()


    # ========================================================
    # M. DRAW SAHA-BOLTZMANN PLOT
    # ========================================================

    plt.figure(
        figsize=(10, 6.5)
    )


    # Regression line + worst-case fit uncertainty band
    E_plot = np.linspace(
        E_all.min() - 1000,
        E_all.max() + 1000,
        500
    )

    Y_plot = (
        m * E_plot
        +
        b
    )

    Y_plot_lower, Y_plot_upper = fit_uncertainty_band(
        E_plot,
        m,
        sigma_m,
        b,
        sigma_b
    )

    plt.fill_between(
        E_plot,
        Y_plot_lower,
        Y_plot_upper,
        alpha=0.15,
        color="gray",
        label="Fit uncertainty (worst case)",
        zorder=1
    )

    plt.plot(
        E_plot,
        Y_plot,
        linewidth=2.5,
        label="16-line fit",
        zorder=2
    )


    # Ca I — propagated line-by-line y-error bars
    plt.errorbar(
        E_CaI,
        Y_CaI,
        yerr=sigma_Y_CaI,
        fmt="o",
        markersize=7,
        capsize=4,
        elinewidth=1.2,
        ecolor="black",
        label=r"Ca I ($\pm\Delta y$, linear relative-error propagation)",
        zorder=3
    )


    # Ca II  — y-error bars from the propagated Saha-correction
    # uncertainty (sigma_saha), calculated from sigma_T and sigma_Ne
    plt.errorbar(
        E_CaII,
        Y_CaII_corrected,
        yerr=sigma_Y_CaII_corrected,
        fmt="s",
        markersize=8,
        capsize=4,
        elinewidth=1.5,
        color="tab:orange",
        ecolor="black",
        label="Ca II — Saha corrected (±Δy, linear propagation)",
        zorder=4
    )


    plt.xlabel(
        r"Upper-level energy $E_k$ (cm$^{-1}$)",
        fontsize=12
    )

    plt.ylabel(
        "Saha-Boltzmann Y-coordinate",
        fontsize=12
    )

    plt.title(
        f"Saha-Boltzmann Plot — Iteration {iteration}",
        fontsize=14
    )


    # Information box away from data
    information = (
        f"Iteration = {iteration}\n"
        f"T = {T_new:.2f} ± {sigma_T_new:.2f} K\n"
        f"Ne = {Ne:.3e} ± {sigma_Ne:.2e} cm$^{{-3}}$\n"
        f"Slope = {m:.3e} ± {sigma_m:.2e} cm\n"
        f"Saha = {saha:.5f} ± {sigma_saha:.5f}\n"
        f"R² = {R2:.5f}"
    )


    plt.text(
        0.02,
        0.97,
        information,
        transform=plt.gca().transAxes,
        ha="left",
        va="top",
        fontsize=9,
        bbox=dict(
            facecolor="white",
            edgecolor="black",
            alpha=0.9
        )
    )


    plt.grid(alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()


    # ========================================================
    # M. CHECK CONVERGENCE
    # ========================================================

    temperature_change = abs(
        T_new - T
    )

    print(
        f"\nTemperature change = "
        f"{temperature_change:.6f} K"
    )


    if temperature_change < tolerance:

        print()
        print("*" * 100)
        print("CONVERGENCE ACHIEVED")
        print("*" * 100)

        print(
            f"Final temperature = "
            f"{T_new:.3f} +/- "
            f"{sigma_T_new:.3f} K"
        )

        print(
            f"Final electron density = "
            f"{Ne:.6e} +/- "
            f"{sigma_Ne:.6e} cm^-3"
        )

        break


    # --------------------------------------------------------
    # Update temperature and uncertainty
    # --------------------------------------------------------

    T = T_new
    sigma_T = sigma_T_new


# ============================================================
# 17. FINAL ITERATION TABLE
# ============================================================

print()
print()
print("=" * 175)
print("FINAL SAHA-BOLTZMANN ITERATION TABLE")
print("=" * 175)


print(
    f"{'Iter':>4} | "
    f"{'T_in':>11} | "
    f"{'σT_in':>10} | "
    f"{'omega':>12} | "
    f"{'Ne':>15} | "
    f"{'σNe':>15} | "
    f"{'Saha':>9} | "
    f"{'σSaha':>9} | "
    f"{'Slope':>14} | "
    f"{'σSlope':>13} | "
    f"{'T_out':>11} | "
    f"{'σT_out':>10}"
)

print("-" * 175)


for row in results:

    (
        iteration,
        T_in,
        sigma_T_in,
        omega,
        sigma_omega,
        Ne,
        sigma_Ne,
        saha,
        m,
        sigma_m,
        T_out,
        sigma_T_out,
        R2,
        sigma_saha
    ) = row


    print(
        f"{iteration:4d} | "
        f"{T_in:11.3f} | "
        f"{sigma_T_in:10.3f} | "
        f"{omega:12.8e} | "
        f"{Ne:15.6e} | "
        f"{sigma_Ne:15.6e} | "
        f"{saha:9.5f} | "
        f"{sigma_saha:9.5f} | "
        f"{m:14.6e} | "
        f"{sigma_m:13.6e} | "
        f"{T_out:11.3f} | "
        f"{sigma_T_out:10.3f}"
    )


print("=" * 175)


# ============================================================
# 18. FINAL RESULT
# ============================================================

final = results[-1]


final_T = final[10]
final_sigma_T = final[11]

final_Ne = final[5]
final_sigma_Ne = final[6]

final_m = final[8]
final_sigma_m = final[9]

final_omega = final[3]
final_sigma_omega = final[4]

final_saha = final[7]
final_sigma_saha = final[13]


print()
print("=" * 100)
print("FINAL RESULTS")
print("=" * 100)

print(
    f"Initial Ca I temperature = "
    f"{T_initial:.3f} +/- "
    f"{sigma_T_initial:.3f} K"
)

print(
    f"Final slope = "
    f"{final_m:.9e} +/- "
    f"{final_sigma_m:.9e} cm"
)

print(
    f"Final temperature = "
    f"{final_T:.3f} +/- "
    f"{final_sigma_T:.3f} K"
)

print(
    f"Final omega = "
    f"{final_omega:.9e} +/- "
    f"{final_sigma_omega:.9e} nm"
)

print(
    f"Final electron density = "
    f"{final_Ne:.6e} +/- "
    f"{final_sigma_Ne:.6e} cm^-3"
)

print(
    f"Final Saha correction = "
    f"{final_saha:.6f} +/- {final_sigma_saha:.6f}"
)

print("=" * 100)


# ============================================================
# 19. UNCERTAINTY SUMMARY
# ============================================================

print()
print("=" * 100)
print("UNCERTAINTY SUMMARY")
print("=" * 100)

print(
    f"Stark width = "
    f"{delta_lambda:.3f} +/- "
    f"{sigma_delta_lambda:.3f} nm"
)

print(
    f"Relative Stark-width uncertainty = "
    f"{sigma_delta_lambda/delta_lambda*100:.3f} %"
)

print(
    f"Final relative slope uncertainty = "
    f"{final_sigma_m/abs(final_m)*100:.3f} %"
)

print(
    f"Final relative temperature uncertainty = "
    f"{final_sigma_T/final_T*100:.3f} %"
)

print(
    f"Final relative electron-density uncertainty = "
    f"{final_sigma_Ne/final_Ne*100:.3f} %"
)

print(
    f"Final Saha relative uncertainty (DeltaS/S) = "
    f"{final_sigma_saha:.5f} "
    f"(common Saha contribution; line-specific Ca II Delta y also includes dI/I, dl/l and dA/A)"
)

print("=" * 100)
```
